### Maximum Absolute Error (MAE)

```python
from sklearn.metrics import max_error
import numpy as np

y_real = np.array([1, 2, 3, 4, 5])
y_pred = np.array([1, 2, 3, 4, -5])

max_error(y_real, y_pred)
```

### Mean Absolute Error (MAE)

```python
from sklearn.metrics import mean_absolute_error
import numpy as np

y_real = np.array([1, 2, 3, 4, 5])
y_pred = np.array([1, 2, 3, 3, -5])

a = y_real - y_pred
b = np.sum(a)
c = b / len(y_real)

mean_absolute_error(y_real, y_pred) == c
```

### Mean Squared Error (MSE)

```python
from sklearn.metrics import mean_squared_error
import numpy as np

y_real = np.array([1, 2, 3, 4, 5])
y_pred = np.array([1, 2, 3, 3, -5])

a = y_real - y_pred
b = a**2
c = np.sum(b)
d = c / y_real.size

mean_squared_error(y_real, y_pred) == d
```

### Residual sum of squares (RSS)

```python
from sklearn.metrics import mean_squared_error
import numpy as np

y_real = np.array([1, 2, 3, 4, 5])
y_pred = np.array([1, 2, 3, 3, -5])

a = y_real - y_pred
b = a**2
c = np.sum(b)
c
```

### Root Mean Squared Error (RMSE)

```python
from sklearn.metrics import root_mean_squared_error
import numpy as np

y_real = np.array([1, 2, 3, 4, 5])
y_pred = np.array([2, 2, 3, 4, -5])

a = y_real - y_pred
b = a**2
c = np.sum(b)
d = c / y_real.size
e = np.sqrt(d)

root_mean_squared_error(y_real, y_pred)
```

```python
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
import numpy as np

X = np.arange(0, 10)
y_real = np.random.normal(0, np.sqrt(X.size), X.size)
y_pred = np.linspace(np.min(y_real), np.max(y_real), X.size)

plt.scatter(X, y_real)
plt.plot(X, y_pred, linestyle="--")
plt.grid(True, alpha=0.3)

r2_score(y_real, y_pred)
```
